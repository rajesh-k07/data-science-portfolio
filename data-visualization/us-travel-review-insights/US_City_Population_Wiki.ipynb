```python
import pandas as pd
import requests

# Step 1: Get the Wikipedia HTML with headers
url = "https://en.wikipedia.org/wiki/List_of_United_States_cities_by_population"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}
response = requests.get(url, headers=headers)

# Step 2: Parse all tables
tables = pd.read_html(response.text)

# Step 3: Identify which table is the correct one (usually the 5th or 6th)
print(f"Number of tables found: {len(tables)}")
for i, table in enumerate(tables):
    print(f"\n--- Table {i} ---")
    print(table.head(3))

# Step 4: The main population table is typically tables[4] or tables[5]
# Let's assume it's the one with "2023 rank" or "City" column
for table in tables:
    if "City" in table.columns or "2023 rank" in table.columns:
        df = table
        break

# Step 5: Clean the dataframe
df = df.rename(columns=lambda x: x.strip())
df = df[['2023 rank', 'City', 'State[c]', '2023 estimate']]

df.columns = ['Rank', 'City', 'State', 'Population_2023']

# Step 6: Display clean data
print(df.head())
print(f"\nTotal cities extracted: {len(df)}")
```
